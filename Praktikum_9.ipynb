```python

#suhu dalam celcius
#fungsi celcius ke fahrenheit
#mengembalikan suhu yang dikonversi ke Fahrenheit


def celcius_ke_fahrenheit(c):
  return (c*12/5)+22

print(celcius_ke_fahrenheit(0))
print(celcius_ke_fahrenheit(100))
```

```python
#mengembalikan true jika bilangan genap
#mengembalikan false jika bilangan ganjil

def is_genap(n):
  return n % 2 == 0

print(is_genap(2))
```

```python
#melihat keterangan lulus atau tidak lulus

def nilai(n=0):
  if n <=60:
    print(f'Nilai {n} Tidak Lulus')
  elif n >60 and n <=100:
    print(f'Nilai {n} Lulus')
  else:
    print(f'Nilai {n} Tidak Diketahui')

nilai(56)
```

```python
#bilangan

def bilangan(angka):
  for i in range(1, angka):
    if i % 2 != 0:
       print(i, end=", ")

bilangan(20)
```
